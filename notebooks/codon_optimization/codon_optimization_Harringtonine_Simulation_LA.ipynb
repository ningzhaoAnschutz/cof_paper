```python
# Align all the amino acid sequences with Biopython
from Bio.Seq import Seq
from Bio.SeqRecord import SeqRecord
from Bio import Align
from Bio.Align import MultipleSeqAlignment
import io

# First, translate all GFP sequences to amino acids
def translate(seq):
    return str(Seq(seq).translate(to_stop=True))

# Create a list of SeqRecord objects for alignment
seq_records = []
for i, dna_seq in enumerate(list_of_GFPs):
    aa_seq = translate(dna_seq)
    seq_record = SeqRecord(Seq(aa_seq), id=list_of_GFP_names[i], description="")
    seq_records.append(seq_record)

# For pairwise alignment comparison of all sequences against the first one (GFPuv)
print("=" * 80)
print("PAIRWISE ALIGNMENTS (comparing each sequence to GFPuv)")
print("=" * 80)

aligner = Align.PairwiseAligner()
aligner.mode = 'global'
aligner.match_score = 2
aligner.mismatch_score = -1
aligner.open_gap_score = -10
aligner.extend_gap_score = -0.5

reference_seq = translate(list_of_GFPs[0])  # GFPuv as reference

for i, dna_seq in enumerate(list_of_GFPs[1:], 1):
    query_seq = translate(dna_seq)
    alignments = aligner.align(reference_seq, query_seq)
    best_alignment = alignments[0]
    
    # Calculate identity
    aligned_ref, aligned_query = str(best_alignment).split('\n')[0], str(best_alignment).split('\n')[2]
    matches = sum(1 for a, b in zip(aligned_ref, aligned_query) if a == b and a != '-')
    total = max(len(aligned_ref.replace('-', '')), len(aligned_query.replace('-', '')))
    identity = (matches / total) * 100
    
    print(f"\n{list_of_GFP_names[0]} vs {list_of_GFP_names[i]}:")
    print(f"  Alignment Score: {best_alignment.score:.1f}")
    print(f"  Identity: {identity:.1f}%")
    print(f"  Sequence lengths: {len(reference_seq)} vs {len(query_seq)} amino acids")

# Display simple alignment comparison
print("\n" + "=" * 80)
print("AMINO ACID SEQUENCES COMPARISON")
print("=" * 80)

# Print all amino acid sequences with their names
for i, dna_seq in enumerate(list_of_GFPs):
    aa_seq = translate(dna_seq)
    print(f"\n{list_of_GFP_names[i]} ({len(aa_seq)} aa):")
    # Print in blocks of 60 amino acids
    for j in range(0, len(aa_seq), 60):
        print(f"  {j+1:4d}: {aa_seq[j:j+60]}")

# Highlight differences between sequences
print("\n" + "=" * 80)
print("SEQUENCE DIFFERENCES (relative to GFPuv)")
print("=" * 80)

reference_aa = translate(list_of_GFPs[0])
for i, dna_seq in enumerate(list_of_GFPs[1:], 1):
    query_aa = translate(dna_seq)
    differences = []
    for pos, (ref, qry) in enumerate(zip(reference_aa, query_aa)):
        if ref != qry:
            differences.append(f"{ref}{pos+1}{qry}")
    
    if differences:
        print(f"\n{list_of_GFP_names[i]}: {len(differences)} differences")
        print(f"  Mutations: {', '.join(differences)}")
    else:
        print(f"\n{list_of_GFP_names[i]}: Identical to GFPuv (synonymous codon changes only)")
```

```python
def build_construct(
    seq: str,
    n: int,
    construct_start_sequence: str,
    construct_end_sequence: str,
    linker_sequence: str
) -> str:
    """
    Build a construct by repeating `seq` n times, inserting `linker_sequence`
    between repeats, and wrapping with start/end sequences.

    Args:
        seq: The core sequence to repeat.
        n: Number of times to include `seq` (must be >=1).
        construct_start_sequence: Prefix sequence.
        construct_end_sequence: Suffix sequence.
        linker_sequence: Sequence to put between repeats.

    Returns:
        The full construct string.
    """
    if n < 1:
        raise ValueError(f"n must be >= 1 (got {n})")

    # Start with the first copy of seq
    body = seq

    # For each additional repeat, prepend a linker then another seq
    for _ in range(1, n):
        body += linker_sequence + seq

    # Wrap with start/end
    return construct_start_sequence + body + construct_end_sequence
```

```python
def simulate_harringtonine(gene_sequence,ki=0.04,ke_global=5, use_sem=False,max_frame=20):

    # reading the gene sequence
    protein, rna, _, indexes_tags, _, _, _  = read_sequence(seq=gene_sequence, min_protein_length=50,TAG=[HA_TAG])
    gene_length = len(protein) # adding 1 to account for the stop codon
    tag_positions_first_probe_vector = indexes_tags[0]
    first_probe_position_vector = create_probe_vector(tag_positions_first_probe_vector, gene_length)

    #ke_global = 5
    #ki = 0.03
    ke_codon_dependent = calculate_codon_elongation_rates (rna, global_elongation_rate=ke_global)

    #max_frame = 15 # minutes
    full_frames = np.arange(0, max_frame) 
    harringtonine_frame = 5 # minutes
    full_frames = full_frames - harringtonine_frame
        
    ########################## Modeling  #########################################
    harrintonine_delay_time_to_enter_cell = 60 # seconds. Accodinf to Tanenbaum paper 2016.
    number_repetitions = 200
    burnin_time = 2000
    #t_max = 360*5 #timePerturbationApplication + 25*60  # Maximum time
    t_max = max_frame*60 #timePerturbationApplication + 25*60  # Maximum time
    step_size_in_sec = 1
    time_array = np.arange(0, t_max, step_size_in_sec)
    #plot_name, rna, first_probe_position_vector, gene_length,list_param = model_and_data_selection(processed_data)
    timePerturbationApplication = harringtonine_frame*60 + harrintonine_delay_time_to_enter_cell
    evaluatingInhibitor = 1
    ssa_array = simulate_TASEP_SSA(ki, 
                                ke_codon_dependent,
                                gene_length, 
                                t_max,
                                time_interval_in_seconds=step_size_in_sec,
                                number_repetitions=number_repetitions, 
                                first_probe_position_vector=first_probe_position_vector, 
                                timePerturbationApplication = timePerturbationApplication,
                                inhibitor_effectiveness=95,
                                evaluatingInhibitor=evaluatingInhibitor,
                                burnin_time=burnin_time,
                                constant_elongation_rate=None, # This ensures that codo-usage is taken into account
                                fast_output=False)[2]
    # downsample time array and simulation output to 1 minute resolution.
    time_array_downsampled = time_array[::60]
    ssa_array_downsampled = ssa_array[:,::60]
    # plotting 
    time_array_min = time_array_downsampled/60
    normalized_data = np.zeros_like(ssa_array_downsampled)
    for i in range(ssa_array_downsampled.shape[0]):
        mean_before_treatment = np.mean(ssa_array_downsampled[i,:harringtonine_frame+1])
        normalized_data[i] = ssa_array_downsampled[i]/mean_before_treatment
    mean_intensity_ssa_ht = np.mean(normalized_data, axis=0)

    if use_sem:
        err_intensity_ssa_ht = np.std(normalized_data, axis=0) / np.sqrt(normalized_data.shape[0])
    else:
        err_intensity_ssa_ht = np.std(normalized_data, axis=0) #/np.sqrt(normalized_data.shape[0])

    return time_array_min, mean_intensity_ssa_ht, err_intensity_ssa_ht

```

```python
def plot_harringtonine(legend_label, time_array_min, mean_intensity_ssa_ht,err_intensity_ssa_ht):
    fig, ax = plt.subplots(figsize=(6, 3), facecolor='white')
    ax.set_facecolor('white')

    # Plotting the model
    plt.plot(time_array_min-5, mean_intensity_ssa_ht,'-',  color='red', linewidth=3,label=legend_label)
    plt.fill_between(time_array_min-5, mean_intensity_ssa_ht - err_intensity_ssa_ht,
                    mean_intensity_ssa_ht + err_intensity_ssa_ht, color='red', alpha=0.1)

    # Plot the harringtonine frame as a vertical line (example at x=0)
    ax.axvline(x=0, color='black', linestyle='--', linewidth=1 ) #, label='Harringtonine')

    # ----------------------------
    # 4) Final plot formatting
    # ----------------------------
    ax.set_xlabel("Time (min)", fontdict={ 'size': 16})
    ax.set_ylabel("Norm. Intensity", fontdict={ 'size': 16})
    ax.tick_params(axis='both', which='major', labelsize=12)
    # set title 
    #ax.set_title(plot_name, fontdict={'family': 'Arial', 'size': 16})

    # # Set spines color to black
    for spine in ax.spines.values():
        spine.set_color('black')
        spine.set_linewidth(1.5)

    plt.ylim(0, 1.4)
    plt.tight_layout()
    ax.legend(fontsize=10)
    #plt.savefig(results_folder.joinpath('HT_'+plot_name+'.png'), dpi=600)
    #plt.savefig(results_folder.joinpath('HT_'+plot_name+'.svg'), dpi=600)

    plt.show()
```

```python
def plot_harringtonine_all(
    legend_labels,    # str or list of str
    time_arrays_min,  # np.ndarray or list of np.ndarray
    mean_intensities, # np.ndarray or list of np.ndarray
    err_intensities   # np.ndarray or list of np.ndarray
):
    # --- normalize legend_labels to list ---
    if not isinstance(legend_labels, (list, tuple)):
        legend_labels = [legend_labels]
    n = len(legend_labels)

    # --- broadcast time_arrays_min if needed ---
    if not isinstance(time_arrays_min, (list, tuple)):
        time_arrays_min = [time_arrays_min] * n

    # --- similarly ensure mean_intensities and err_intensities are lists ---
    if not isinstance(mean_intensities, (list, tuple)):
        mean_intensities = [mean_intensities] * n
    if not isinstance(err_intensities, (list, tuple)):
        err_intensities = [err_intensities] * n

    # --- now all four lists have length n ---
    assert len(time_arrays_min) == n == len(mean_intensities) == len(err_intensities), \
        "All four inputs must be lists of the same length."

    # 1) create figure
    fig, ax = plt.subplots(figsize=(6, 3), facecolor='white')
    ax.set_facecolor('white')

    for label, t_min, mean_ht, err_ht in zip(
        legend_labels, time_arrays_min, mean_intensities, err_intensities
    ):
        t_plot = np.array(t_min) - 5
        ax.plot(t_plot, mean_ht, '-', linewidth=2, label=label)
        ax.fill_between(t_plot,
                        mean_ht - err_ht,
                        mean_ht + err_ht,
                        alpha=0.05)

    ax.axvline(x=0, linestyle='--', color='k', linewidth=1)
    ax.set_xlabel("Time (min)", size=16)
    ax.set_ylabel("Norm. Intensity", size=16)
    for spine in ax.spines.values():
        spine.set_linewidth(1.5)
    ax.set_ylim(0, 1.5)
    ax.legend(fontsize=10, frameon=False)
    plt.tight_layout()
    plt.show()
```
