# align all the amino acid sequences with biopython




```python
# Align all the amino acid sequences with Biopython
from Bio.Seq import Seq
from Bio.SeqRecord import SeqRecord
from Bio import Align
from Bio.Align import MultipleSeqAlignment
import io

# First, translate all GFP sequences to amino acids
def translate(seq):
    return str(Seq(seq).translate(to_stop=True))

# Create a list of SeqRecord objects for alignment
seq_records = []
for i, dna_seq in enumerate(list_of_GFPs):
    aa_seq = translate(dna_seq)
    seq_record = SeqRecord(Seq(aa_seq), id=list_of_GFP_names[i], description="")
    seq_records.append(seq_record)

# For pairwise alignment comparison of all sequences against the first one (GFPuv)
print("=" * 80)
print("PAIRWISE ALIGNMENTS (comparing each sequence to GFPuv)")
print("=" * 80)

aligner = Align.PairwiseAligner()
aligner.mode = 'global'
aligner.match_score = 2
aligner.mismatch_score = -1
aligner.open_gap_score = -10
aligner.extend_gap_score = -0.5

reference_seq = translate(list_of_GFPs[0])  # GFPuv as reference

for i, dna_seq in enumerate(list_of_GFPs[1:], 1):
    query_seq = translate(dna_seq)
    alignments = aligner.align(reference_seq, query_seq)
    best_alignment = alignments[0]
    
    # Calculate identity
    aligned_ref, aligned_query = str(best_alignment).split('\n')[0], str(best_alignment).split('\n')[2]
    matches = sum(1 for a, b in zip(aligned_ref, aligned_query) if a == b and a != '-')
    total = max(len(aligned_ref.replace('-', '')), len(aligned_query.replace('-', '')))
    identity = (matches / total) * 100
    
    print(f"\n{list_of_GFP_names[0]} vs {list_of_GFP_names[i]}:")
    print(f"  Alignment Score: {best_alignment.score:.1f}")
    print(f"  Identity: {identity:.1f}%")
    print(f"  Sequence lengths: {len(reference_seq)} vs {len(query_seq)} amino acids")

# Display simple alignment comparison
print("\n" + "=" * 80)
print("AMINO ACID SEQUENCES COMPARISON")
print("=" * 80)

# Print all amino acid sequences with their names
for i, dna_seq in enumerate(list_of_GFPs):
    aa_seq = translate(dna_seq)
    print(f"\n{list_of_GFP_names[i]} ({len(aa_seq)} aa):")
    # Print in blocks of 60 amino acids
    for j in range(0, len(aa_seq), 60):
        print(f"  {j+1:4d}: {aa_seq[j:j+60]}")

# Highlight differences between sequences
print("\n" + "=" * 80)
print("SEQUENCE DIFFERENCES (relative to GFPuv)")
print("=" * 80)

reference_aa = translate(list_of_GFPs[0])
for i, dna_seq in enumerate(list_of_GFPs[1:], 1):
    query_aa = translate(dna_seq)
    differences = []
    for pos, (ref, qry) in enumerate(zip(reference_aa, query_aa)):
        if ref != qry:
            differences.append(f"{ref}{pos+1}{qry}")
    
    if differences:
        print(f"\n{list_of_GFP_names[i]}: {len(differences)} differences")
        print(f"  Mutations: {', '.join(differences)}")
    else:
        print(f"\n{list_of_GFP_names[i]}: Identical to GFPuv (synonymous codon changes only)")
```
