# Laser Intensity Extraction from LIF Files

This notebook extracts laser intensity metadata from all LIF files in a specified directory.
The results are exported as a CSV file for analysis in Excel.

````python
from pathlib import Path
import sys

# Add repo root to path so we can import utilities
current_dir = Path().resolve()
repo_root = current_dir.parents[1]  # image_conditions/ -> notebooks/ -> cof_paper/
sys.path.insert(0, str(repo_root))

from microlive.imports import *
from microlive import microscopy as mi
one_drive_dir = mi.Utilities.get_one_drive_dir()

# Import extract_laser_intensities from utilities.metadata
from utilities import extract_laser_intensities
````

## Function imported from `utilities.metadata`

The `extract_laser_intensities` function is now available from the shared utilities package:
```python
from utilities import extract_laser_intensities
```

````python

data_dir_0 = Path('/Volumes/Luis_DRIVE/CoF Manuscript LIFs/CoF AC/GFPuv')
data_dir_1 = Path('/Volumes/Luis_DRIVE/CoF Manuscript LIFs/CoF AC/sfGFP')
data_dir_2 = Path('/Volumes/Luis_DRIVE/CoF Manuscript LIFs/Fig 4_fast Xbp1u ACF LIF Files/pRS038')
data_dir_3 = Path('/Volumes/Luis_DRIVE/CoF Manuscript LIFs/Fig 4_fast Xbp1u ACF LIF Files/pRS048')


list_data_dirs = [data_dir_0,data_dir_1,data_dir_2,data_dir_3]

# Create output folder ONCE (outside the loop)
laser_intensities_folder = current_dir / 'laser_intensities'
laser_intensities_folder.mkdir(exist_ok=True)

# Process each data directory
for data_dir in list_data_dirs:
    # File output name based on the data_dir name (spaces replaced with underscores)
    output_file = laser_intensities_folder / (data_dir.name.replace(' ', '_') + '.csv')
    
    # Extract laser intensities
    df_intensities = extract_laser_intensities(data_dir)
    
    # Save to CSV
    df_intensities.to_csv(output_file, index=False)
    print(f'\nResults saved to: {output_file}')
````
