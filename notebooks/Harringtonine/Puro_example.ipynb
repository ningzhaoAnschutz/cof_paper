```python
from pathlib import Path
current_dir = Path().resolve()
from microlive.imports import *
from microlive import microscopy as mi
main_dir = Path().resolve().parent
one_drive_dir = mi.Utilities.get_one_drive_dir()
import Harringtonine as HT
```

```python
data_dir = one_drive_dir / 'General - Zhao (NZ) Lab' / 'Zhao lab shared folder' / 'Our papers' / 'co-translational folding paper' / 'Fig 1' / 'Fig 1_6xGFP puro graphs'
```

```python
# Define the parameters
use_sem = True
show_individual_trajectories = False
show_summary = False
max_percentage_threshold_after_treatment = None  
threshold_percentage_for_runoff = 0.2
frame_interval_sec = 20
results_folder = None
```

```python
harringtonine_frame_index = 4
```

```python

substring_in_data_dir = 'pNZ212' # pNZ212 pNZ370 pNZ381
selected_field = 'spot_int_ch_1' #'spot_int_ch_0' "spot_int_ch_1'
plot_name = 'Nascent Chains'
responding_indices_ch1,frames_experimental_min_recentered_ch1, intensities_normalized_ch1, _, _, _, _ = HT.process_HT_data(data_dir, harringtonine_frame_index, substring_in_data_dir, selected_field, use_sem, show_summary, max_percentage_threshold_after_treatment,frame_interval_sec=frame_interval_sec )
HT.plot_harringtonine(frames_experimental_min_recentered_ch1, intensities_normalized_ch1, harringtonine_frame_index, show_runoff_time = True, plot_name=plot_name,threshold_percentage_for_runoff=threshold_percentage_for_runoff, figsize=(8,5))
```
